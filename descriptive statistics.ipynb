```python
import pandas as pd
import statistics
import numpy as np

# Declaration
# Descriptive Statistic for Analyzed Accidents by Day of the Week and Crash Date
data = pd.read_csv("traffic_accidents.csv")

# Convert Crash Day
data['crash_date'] = pd.to_datetime(data['crash_date'], format='%m/%d/%Y %I:%M:%S %p')
data['Crash Day'] = data['crash_date'].dt.day_name()

# Unique values for crash day
crash_days = data['Crash Day'].unique().tolist()

results = []
for day in crash_days:
    day_data = data[data['Crash Day'] == day]
    
    total_accident = day_data['injuries_total'].count() 
    min_injury = day_data['injuries_total'].min()  
    max_injury = day_data['injuries_total'].max()  
    avg_injury = day_data['injuries_total'].mean()  
    
    # Append result dictionary for each crash day
    results.append({
        'Crash Day': day,
        'Total Accident': total_accident,
        'Min Injuries Total': min_injury,
        'Max Injuries Total': max_injury,
        'Average Injuries Total': round(avg_injury, 2),
    })

# Sort
res_df = pd.DataFrame(results)
res_df = res_df.sort_values(by='Total Accident', ascending=False)
res_df
```

```python
import pandas as pd
import statistics
import numpy as np

# Declaration
# Descriptive Statistic for Analyzed Accidents by Day Name and Is Injured
data = pd.read_csv("traffic_accidents.csv")

# Convert Crash Date
data['crash_date'] = pd.to_datetime(data['crash_date'], format='%m/%d/%Y %I:%M:%S %p')

# Extract Day Name
data['Day Name'] = data['crash_date'].dt.day_name()

# Categorize Injured
def categorize_injured(val):
    if val == 0:
        return "No Injured"
    else:
        return "Injured"

# Group by Injured and Count
data['is_injured'] = data['injuries_total'].apply(categorize_injured)

# Combine Day Name and Is Injured
data['Day-Is Injured'] = data['Day Name'] + ' - ' + data['is_injured']

# Unique values for Day-Is Injured
day_time_groups = data['Day-Is Injured'].unique().tolist()

results = []
for group in day_time_groups:
    group_data = data[data['Day-Is Injured'] == group]
    
    total_accident = group_data['injuries_total'].count() 
    
    # Append result dictionary for each Day-Is Injured
    results.append({
        'Day-Is Injured': group,
        'Total Accident': total_accident
    })

# Sort
res_df = pd.DataFrame(results)
res_df = res_df.sort_values(by='Total Accident', ascending=False)
res_df
```

```python
import pandas as pd
import statistics
import numpy as np

# Declaration
# Descriptive Statistic for Analyzed Accidents by Time of Day and Is Intersection Related
data = pd.read_csv("traffic_accidents.csv")

# Extract Day and Hour from crash_date
data['crash_date'] = pd.to_datetime(data['crash_date'], format='%m/%d/%Y %I:%M:%S %p')
data['day'] = data['crash_date'].dt.day_name()
data['hour'] = data['crash_date'].dt.hour

# Categorize hour into time periods
def categorize_time(hour):
    if 0 <= hour <= 5:
        return "Midnight"
    elif 6 <= hour <= 11:
        return "Morning"
    elif 12 <= hour <= 17:
        return "Evening"
    elif 18 <= hour <= 23:
        return "Night"

data['time'] = data['hour'].apply(categorize_time)

# Combine Time and Is Intersection Related
data['Time-Is Intersection Related'] = data['time'] + ' - ' + data['intersection_related_i']

# Unique values for Time-Is Intersection Related
day_time_groups = data['Time-Is Intersection Related'].unique().tolist()

results = []
for group in day_time_groups:
    group_data = data[data['Time-Is Intersection Related'] == group]
    
    total_accident = group_data['injuries_total'].count() 
    
    # Append result dictionary for each Day-Is Injured
    results.append({
        'Day-Is Intersection Related': group,
        'Total Accident': total_accident
    })

# Sort
res_df = pd.DataFrame(results)
res_df = res_df.sort_values(by='Total Accident', ascending=False)
res_df
```

```python
import pandas as pd
import statistics
import numpy as np

# Declaration
# Descriptive Statistic for Analyzed Accidents by Weather Condition, Roadway Surface Condition and Is Intersection Related
data = pd.read_csv("traffic_accidents.csv")

target_col = ['roadway_surface_cond','weather_condition']

for dt in target_col:
    col_clean = dt.replace("_"," ").title()
    # Combine Weather Condition, Roadway Surface Condition and Is Intersection Related
    data[f'{col_clean}-Is Intersection Related'] = data[dt] + ' - ' + data['intersection_related_i']
    
    # Unique values for Weather Condition, Roadway Surface Condition-Is Intersection Related
    day_time_groups = data[f'{col_clean}-Is Intersection Related'].unique().tolist()
    
    results = []
    for group in day_time_groups:
        group_data = data[data[f'{col_clean}-Is Intersection Related'] == group]
        
        total_accident = group_data['injuries_total'].count() 
        
        # Append result dictionary for each Weather Condition, Roadway Surface Condition-Is Injured
        results.append({
            f'{col_clean}-Is Intersection Related': group,
            'Total Accident': total_accident
        })
    
    # Sort
    res_df = pd.DataFrame(results)
    res_df = res_df.sort_values(by='Total Accident', ascending=False)
    print(res_df)
    print("\n")
```
